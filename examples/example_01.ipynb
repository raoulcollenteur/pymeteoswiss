```python
from pathlib import Path
import pymeteoswiss as pms

%load_ext autoreload
%autoreload 2
```

## 1. Set the datapaths

```python
path = Path("data/")

if not path.exists():
    raise FileNotFoundError(
        f"Data directory '{path}' does not exist. Please create it"
        "and add the necessary data files."
    )
```

## 2. Load the precipitation time series for one location

```python
x, y = 2527330, 1163750
pms.read_data(path, variable="RhiresD", x=x, y=y)
```

## 3. Load the time series for two locations

```python
x, y = [2527330, 2727330], [1163750, 1303750]
df = pms.read_data(path, variable="RhiresD", x=x, y=y)
df
```

## Plot the time series

```python
df.plot(
    figsize=(10, 3),
    title="RhiresD at multiple locations",
    ylabel="RhiresD (mm/d)",
    drawstyle="steps-post",
);
```
